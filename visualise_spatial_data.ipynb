```python
import numpy as np
import matplotlib.pyplot as plt

from plotting_helper import *
```

```python
dat = np.loadtxt('rib_dat_no_amino.gdat')

observables = ['t', 'R', 'p_l', 'xi_l', 'xr_l', 'xt_l', 'p_s', 'xi_s', 'xr_s', 'xt_l']
```

```python
layout = '''
    AABBC
    AABBC
    '''

fig = plt.figure(figsize=(16, 8))

height_ratios = [1,0.1]
width_ratios = [1,0.1,1,0.1,1]


specs, gs = panel_specs(layout, fig=fig)
gs.set_height_ratios(height_ratios)
gs.set_width_ratios(width_ratios)
gs.hspace=0.8
axes = {}
for letter in 'ABC':
    axes[letter] = ax = fig.add_subplot(specs[letter])
label_panels(axes, letters='ABC', postfix='', offset_left=1.2)
for a in list('ABC'):
    axes[a].set_axis_off()
    
    
ax = fig.add_subplot(gs[0,0])
ax.plot(dat[:,0], dat[:,1])
ax.set_xlabel('t')
ax.set_ylabel('R')
ax.set_ylim(-1,np.max(dat[:,1]))
clear_axes(ax)

#max for y axis
maxy = np.max(dat[:,2:])

ax = fig.add_subplot(gs[0,2])
ax.plot(dat[:,0], dat[:,2], label = 'P')
ax.plot(dat[:,0], dat[:,3], label = 'initiated')
ax.plot(dat[:,0], dat[:,4], label = 'elongated')
ax.plot(dat[:,0], dat[:,5], label = 'terminating')
ax.set_xlabel('t')
ax.set_ylabel('#')
ax.set_title('long')
ax.set_ylim(-1, maxy)
clear_axes(ax)

ax = fig.add_subplot(gs[0,4])
ax.plot(dat[:,0], dat[:,6], label = 'P')
ax.plot(dat[:,0], dat[:,7], label = 'initiated')
ax.plot(dat[:,0], dat[:,8], label = 'elongated')
ax.plot(dat[:,0], dat[:,9], label = 'terminating')
ax.set_xlabel('t')
ax.set_ylabel('#')
ax.set_title('short')
ax.set_ylim(-1, maxy)
ax.legend(bbox_to_anchor=(1.0, 1.0), loc='upper left',  frameon = False)
clear_axes(ax)

plt.show()
```
