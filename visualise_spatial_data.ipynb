```python
import numpy as np
import matplotlib.pyplot as plt

from plotting_helper import *

dat_dir = 'simulation_results/'
```

```python

R,Y,N = 100,1000,50
l = ['2',10,25]
condition = str(R)+'ribos_'+str(Y)+'aminos_'+str(l)+'_'+str(N)
dat = np.loadtxt(dat_dir+condition+'.gdat')

#Molecules r Ribo(N)
    
#Molecules p_s P(s~0)
#Molecules r_initiated_s nucl(R!1, i~1, s~0).Ribo(N!1)
#Molecules r_elongating_s nucl(R!1, t~0, i~0, s~0).Ribo(N!1)
#Molecules r_terminating_s nucl(R!1, t~1, i~0, s~0).Ribo(N!1)
#Molecules p_l P(s~1)
#Molecules r_initiated_l nucl(R!1, i~1, s~1).Ribo(N!1)
#Molecules r_elongating_l nucl(R!1, t~0, i~0, s~1).Ribo(N!1)
#Molecules r_terminating_l nucl(R!1, t~1, i~0, s~1).Ribo(N!1)

observables = ['t', 'R', 'Y', 'p_s', 'p_l', 'xi_s', 'xi_l' , 'xr_s', 'xr_l', 'xt_s', 'xt_l']
```

```python
layout = '''
    AABBC
    AABBC
    '''

fig = plt.figure(figsize=(16, 8))

height_ratios = [1,0.1]
width_ratios = [1,0.1,1,0.1,1]


specs, gs = panel_specs(layout, fig=fig)
gs.set_height_ratios(height_ratios)
gs.set_width_ratios(width_ratios)
gs.hspace=0.8
axes = {}
for letter in 'ABC':
    axes[letter] = ax = fig.add_subplot(specs[letter])
label_panels(axes, letters='ABC', postfix='', offset_left=1.2)
for a in list('ABC'):
    axes[a].set_axis_off()
    
    
ax = fig.add_subplot(gs[0,0])
ax.plot(dat[:,0], dat[:,1], label = 'R')
ax.plot(dat[:,0], dat[:,2], label = 'Y')
ax.set_xlabel('t')
ax.set_ylabel('# Resources')
ax.set_ylim(-1,np.max(dat[:,2]))
ax.legend(bbox_to_anchor=(0.5,1.0), loc='upper left',  frameon = False)
clear_axes(ax)

#max for y axis
maxy = np.max(dat[:,3:])

ax = fig.add_subplot(gs[0,2])
ax.plot(dat[:,0], dat[:,3], label = 'P')
ax.plot(dat[:,0], dat[:,5], label = 'initiated')
ax.plot(dat[:,0], dat[:,7], label = 'elongated')
ax.plot(dat[:,0], dat[:,9], label = 'terminating')
ax.set_xlabel('t')
ax.set_ylabel('#')
ax.set_title('short')
ax.set_ylim(-1, maxy)
clear_axes(ax)

ax = fig.add_subplot(gs[0,4])
ax.plot(dat[:,0], dat[:,4], label = 'P')
ax.plot(dat[:,0], dat[:,6], label = 'initiated')
ax.plot(dat[:,0], dat[:,8], label = 'elongated')
ax.plot(dat[:,0], dat[:,10], label = 'terminating')
ax.set_xlabel('t')
ax.set_ylabel('#')
ax.set_title('long')
ax.set_ylim(-1, maxy)
ax.legend(bbox_to_anchor=(1.0, 1.0), loc='upper left',  frameon = False)
clear_axes(ax)

#plt.tight_layout()

plt.show()
```
